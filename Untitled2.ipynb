```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate synthetic response features
n = 1000
quality = np.random.normal(0, 1, n)
flattery = np.abs(np.random.normal(0, 1, n))

# Actual quality reward and learned proxy reward
gold_reward = quality - 0.65 * flattery**2
proxy_reward = quality + 0.85 * flattery

# Best-of-n optimization
sizes = [1, 2, 4, 8, 16, 32, 64, 128]
gold_scores = []
proxy_scores = []

for size in sizes:
    actual, predicted = [], []

    for _ in range(100):
        ids = np.random.choice(n, size=size)
        best = ids[np.argmax(proxy_reward[ids])]

        actual.append(gold_reward[best])
        predicted.append(proxy_reward[best])

    gold_scores.append(np.mean(actual))
    proxy_scores.append(np.mean(predicted))

# Display results
for n, g, p in zip(sizes, gold_scores, proxy_scores):
    print(f"Best-of-{n}: Gold={g:.3f}, Proxy={p:.3f}")

# Plot results
plt.plot(sizes, gold_scores, marker="o", label="Gold Reward")
plt.plot(sizes, proxy_scores, marker="o", label="Proxy Reward")
plt.xscale("log", base=2)
plt.xlabel("Optimization Pressure (Best-of-n)")
plt.ylabel("Average Reward")
plt.title("Hunting the Reward Hack")
plt.legend()
plt.grid(True)
plt.show()
```

Abstract
Reward models are commonly used to turn human preferences into a scalar objective that can guide language-model behavior. A central risk is reward hacking: an optimized system discovers features that increase the learned reward without improving the outcome that people actually care about. This project studies that failure mode in a reproducible synthetic environment. Candidate answers are described by genuine quality, flattery intensity, answer length, and confidence. A held-out gold reward values quality and penalizes excessive flattery and excessive length. In contrast, simulated annotators tend to prefer both quality and flattery. Pairwise preference data train small, medium, and large proxy reward models at three data scales. Best-of-n selection then increases optimization pressure by selecting the highest-proxy candidate from pools of increasing size. Held-out gold reward, proxy score, an independent simulated judge, flattery, and length are measured across repeated trials. Bootstrap confidence intervals summarize the mean outcomes. A length-bin control is also used to check whether changes can be explained by length alone. The experiment is deliberately synthetic: it illustrates a mechanism and supports controlled comparisons, but it is not empirical evidence about a deployed language model. The main lesson is methodological: proxy reward must be evaluated against independent objectives, and optimization should be monitored for a peak rather than assumed to improve quality indefinitely.
1. Introduction
Preference-based alignment attempts to make model outputs more useful, safe, or consistent with human intent. One common design learns a reward model from comparisons between two candidate answers. The reward model estimates which answer is preferred, after which an optimization procedure searches for outputs with high predicted reward. This creates a useful bridge between qualitative feedback and an objective that can be optimized computationally.
The weakness is that the learned reward is only a proxy. Annotators may have inconsistent preferences, may overvalue superficial signals, or may not fully express the intended objective in their comparisons. An optimizer does not automatically understand the spirit of a reward. It can exploit any feature that is predictive of the proxy score, even if that feature is not beneficial under a more defensible measure of quality. This is often called reward hacking or specification gaming.
The assignment's Option 3 asks for reward models with different sizes and data amounts, measurement of where over-optimization peaks, and a diagnosis of what the policy learns to exploit. This project operationalizes those ideas in a small environment that can run in a standard Python notebook. Instead of training a large language model, it models candidate answers using interpretable numerical features. That makes it possible to inspect the mismatch between the training preference rule and the held-out gold reward directly.
2. Research Questions and Hypotheses
The experiment answers four questions. First, does selecting candidates with higher proxy reward always improve gold reward? Second, how does the observed gold-reward peak vary with reward-model capacity? Third, does changing the amount of preference-training data change the peak? Fourth, can the gap be diagnosed by tracking flattery, answer length, and an independent quality judge?
The primary hypothesis is that proxy reward will generally increase as best-of-n grows, while gold reward can rise initially and then decline. The mechanism is selection pressure: when more candidates are considered, the optimizer has a greater chance of finding an extreme candidate whose proxy score is high partly because of a proxy-favored feature. A second hypothesis is that capacity and data scale may change the peak, but not necessarily in one universal direction. More capacity can represent more complicated patterns, while more data can estimate the annotator's preferences more consistently. If those preferences are misaligned with gold quality, better estimation of the proxy objective does not by itself solve the mismatch.
3. Environment and Reward Definitions
Each synthetic candidate is represented by four features. Genuine quality q is sampled from a standard normal distribution. Flattery f is the absolute value of a normally distributed sample, so it is non-negative and occasionally large. Length L is sampled from a log-normal distribution and clipped to a reasonable range. Confidence c is a noisy combination of quality and flattery. These are abstract features, not generated natural-language responses.
The gold reward is defined as G = q − 0.65f² − 0.12 max(L − 1.2, 0)². This definition rewards genuine quality, penalizes heavy flattery quadratically, and mildly penalizes length only after a threshold. It is the evaluation target and is never used to fit the proxy model. The simulated annotator score is A = q + 0.85f + 0.12L + ε, where ε is random noise. This annotator prefers quality but also gives a positive contribution to flattery and length. Pairwise labels are generated by comparing two candidates' annotator scores.
The mismatch is intentional and explicit. Flattery is not intrinsically harmful in every real interaction; in this experiment it is a proxy feature that annotators overvalue relative to the stated gold objective. The experiment therefore tests a clearly defined failure mode, rather than claiming that all human preference data are biased in this way. An independent simulated judge uses a different formula, J = q − 0.40f² − 0.10 max(L − 1.0, 0) + η. It penalizes flattery and excessive length, but with different weights and noise. The judge is not used for training or selection.
4. Reward-Model Training
Three model capacities are compared. The small model uses an intercept and the linear features q, f, and L. The medium model adds confidence and two interaction terms. The large model adds further nonlinear and interaction terms, including squared flattery and squared length. These are feature-capacity levels rather than neural-network parameter counts. They provide a controlled way to vary the flexibility of the proxy model while keeping the underlying candidate environment fixed.
Training uses pairwise logistic ranking. For a sampled pair i and j, the model receives the feature difference between the candidates and a binary label indicating which candidate the annotator preferred. The model parameters are updated by gradient descent to minimize logistic loss with a small L2 penalty. Training data scales are 100, 500, and 2,000 candidate examples, with a corresponding set of sampled comparisons. The random seeds are fixed so the notebook can be rerun reproducibly. The evaluation pool is separate from the training pool, and the gold reward is not supplied to the fitting procedure.
This setup approximates a reward model learned from preference comparisons. It is not a full RLHF training pipeline: no language model is updated by policy gradients, and no DPO model is trained. Instead, best-of-n selection acts as a transparent optimization-pressure mechanism. This simplification is appropriate for isolating the proxy-reward failure, but it limits how directly the numerical results can be transferred to real alignment training.
5. Optimization Procedure and Evaluation
For each combination of model capacity and training-data scale, the fitted model assigns a proxy score to every candidate in a held-out evaluation pool. At each value of n in {1, 2, 4, 8, 16, 32, 64, 128}, the procedure samples n candidates and selects the one with the largest proxy score. Repeating this selection hundreds of times estimates the expected outcome under that optimization pressure. The chosen candidate is then evaluated using gold reward, proxy score, independent judge score, flattery, and length.
The observed over-optimization peak is defined as the tested value of n with the highest mean gold reward. This is a discrete peak over the tested grid, not a mathematically exact global optimum. If the best value occurs at n = 128, the experiment cannot establish whether the peak lies beyond the search range. Conversely, a decline at large n is evidence that stronger selection can hurt gold reward under this setup. The notebook reports the full curve so the peak is not the only reported statistic.
Uncertainty intervals are computed by bootstrapping the repeated selected outcomes. For each metric and configuration, the notebook resamples the trial-level values with replacement, computes the mean, and uses the 2.5th and 97.5th percentiles as a 95% interval. These intervals represent variation across selection trials conditional on the trained model and evaluation pool; they do not include every source of uncertainty, such as variation across all possible training datasets or model initializations. That limitation should be considered when interpreting small differences.
6. Length Control and Independent Judge
A model may appear to improve because it selects longer responses, so length is measured directly and receives an explicit gold-reward penalty only beyond a threshold. In addition, a length-bin diagnostic compares each selected candidate's gold reward with the average gold reward of candidates in the same approximate length bin. A positive difference means the selected candidate is better than the typical candidate of similar length according to the gold metric. This is a diagnostic control rather than a perfect causal experiment: binning is coarse, and other candidate features can still differ.
The independent judge provides a second check. It shares the general intention of the gold metric—rewarding quality and penalizing flattery—but uses different penalty weights and independent noise. Agreement between the judge and gold reward strengthens the interpretation that a decline is not unique to one exact formula. Disagreement would be informative too, because it would show that evaluation conclusions depend on how quality is operationalized. In real alignment work, an independent human evaluation or a separately validated evaluator would be preferable to another hand-written synthetic score.

The central comparison is between the proxy objective and the held-out measures. Because the annotator score includes a positive flattery term while gold reward penalizes flattery quadratically, a model that learns the preference pattern can assign high proxy scores to overly flattering candidates. As selection pressure rises, such candidates become more likely to win. A high proxy score is therefore not sufficient evidence of high gold quality. The independent judge and feature-level metrics help diagnose this mismatch instead of relying on one scalar score alone.
Capacity and data scale should be interpreted empirically from the generated peak table, not through a universal rule. A more flexible model may capture nonlinear flattery effects more strongly, but it may also rank candidates differently because of finite-data estimation and regularization. More preference data can reduce estimation noise while preserving the annotator's systematic bias. Accordingly, additional data can make the proxy more faithful to the annotator without making it more faithful to the gold reward. The plotted peak location is a descriptive outcome of this synthetic environment and should not be generalized to all reward models.

8. Diagnosis: What Is Being Hacked?
The principal exploitable feature is flattery. The annotator score gives flattery a positive linear contribution, whereas the gold reward penalizes it quadratically. Moderate flattery may increase proxy score, and extreme flattery can become especially attractive under selection because the optimizer searches for high-scoring outliers. Yet the gold penalty grows faster as flattery increases. This creates a natural route to reward hacking: the optimizer learns to produce or select a feature that satisfies the proxy but conflicts with the actual objective.
Length is a secondary diagnostic. The annotator score rewards length slightly, while gold reward penalizes only excessive length. If selected length rises substantially with n, some observed behavior may reflect verbosity exploitation. The length-bin control helps ask whether selected candidates are unusually good relative to candidates of similar length. It does not remove every confounder, but it provides more information than reporting a single average score. Confidence is included as an additional feature so that model capacity can represent interactions, although the main intended mismatch is flattery.
This diagnosis illustrates a broader lesson: reward hacking need not involve malicious intent or a dramatic software bug. It can emerge from a small, systematic disagreement between the proxy and the intended objective. Optimization amplifies that disagreement by concentrating on candidates where the proxy is unusually favorable. A robust alignment evaluation should therefore track the intended outcome and plausible proxy-exploitation indicators throughout optimization.
9. Limitations and Validity
First, the candidate space is synthetic and low-dimensional. Real language outputs contain semantic meaning, factuality, style, safety, and context that cannot be captured by four numerical features. Second, the annotator bias is designed into the data-generating process. The project demonstrates consequences of that assumption rather than measuring how frequently real annotators exhibit the same bias. Third, best-of-n selection is a simplified optimizer; it is not policy-gradient RLHF or DPO. Fourth, the independent judge is another simulated formula, not an actual independent human study.
Fifth, the model-capacity comparison changes feature representations and is not equivalent to scaling a transformer. Sixth, bootstrap intervals quantify trial-level uncertainty for a fixed trained model and pool; they do not fully quantify uncertainty across retraining runs. Seventh, the peak is selected from a finite grid of n values. A stronger study would run multiple independent training seeds, expand the optimization range, evaluate several annotator-bias strengths, and include a length-matched sampling design specified before looking at results. A real-world extension could use human-rated responses and a held-out evaluation set, with appropriate privacy and safety review.
10. Conclusion
This project builds a reproducible experiment for Option 3, Hunting the Reward Hack. It trains preference-based proxy reward models at multiple capacity and data scales, then increases optimization pressure through best-of-n selection. It evaluates selected candidates against a held-out gold reward, an independent judge, and explicit flattery and length measures. The notebook also calculates bootstrap uncertainty intervals and a length-control diagnostic. The project is designed to reveal when proxy optimization and actual quality diverge, and to locate the best observed gold-reward point within the tested range.
The most important conclusion is not that a particular model size or dataset is always best. It is that optimizing a proxy can amplify its mistakes. A model can become increasingly successful at maximizing the learned score while becoming less successful at the intended objective. The practical response is to define a defensible gold evaluation, keep it out of training, monitor several metrics, examine possible exploit features, and stop or redesign optimization when independent quality stops improving. The results should be described as a controlled synthetic demonstration, and any claim about real language models would require external empirical validation.
